```python
Grd=int(input("enter your grade"))
if Grd>90 and Grd<=100:
    print("A")
elif Grd>80 and Grd<=90:
    print("B")
elif Grd>70 and Grd<=80:
    print("C")
elif Grd>60 and Grd<=70:
    print("D")
else:
    print("F")
```

```python
num=int(input("enter a number"))
count=0
sum=0
for i in range(1,num+1):
    if i%2==0:
        sum+=i
    if i%3==0 and i%5==0:
        count+=1
        print(" 3 5")
    elif i%3==0:
        count+=1
        print(" 3")
    elif i%5==0:
        print(" 5")
    else:
        print(i)
    
print(count)
print(sum)
    
```
